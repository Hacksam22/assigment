```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Labels:", digits.target_names)
plt.figure(figsize=(4, 4))
plt.imshow(digits.images[0], cmap='pink')
plt.title(f"Label: {digits.target[0]}")
plt.axis("off")
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
fig, axes = plt.subplots(2, 5, figsize=(10, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='pink')
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')
plt.tight_layout()
plt.show()
```

Question 2: Write a Python program to display:
a) Number of rows and columns
b) Column names
c) Data types

```python
from sklearn.datasets import load_iris
import pandas as pd
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
print("\nColumn Names:")
print(df.columns)
print("\nData Types:")
print(df.dtypes)
```

Question 3: Write a Python program to count the number of samples available in the Iris dataset and Digits dataset.

```python
from sklearn.datasets import load_iris, load_digits
iris = load_iris()
digits = load_digits()
print("Number of samples in Iris dataset:", len(iris.data))
print("Number of samples in Digits dataset:", len(digits.data))
```
